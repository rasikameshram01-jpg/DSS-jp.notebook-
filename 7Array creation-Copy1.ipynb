```python
fruits=['apple','banana','mango','apple',6]
if"apple"in fruits:
    print("Yes")
else:
    print("No")
```

```python
#List oprations

fruits.append('orange')
print('After reversal:',fruits)

fruits.reverse()
print('After reversal:',fruits)

print(fruits.count('apple'))

fruits.insert(1,'grapes')
print('After insert:',fruits)

fruits.pop(2)
print('After removing third element:',fruits)
```

```python
#Out of the box

companies=("Google","Microsoft","Amazon","TCS")

temp=list(companies)
temp.append("Facebook")
temp.pop(3)
companies=tuple(temp)
print(companies)
print(type(companies))
```

```python
#Mathematical set oprations

A={1,2,3,4}
B={3,4,5,6}

print('Union:',A.union(B))
print('Intersection:',A.intersection(B))
print(A.isdisjoint(B))
print(A.issuperset(B))
```
